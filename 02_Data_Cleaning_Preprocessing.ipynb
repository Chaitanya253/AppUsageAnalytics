# 02 - Data Cleaning and Preprocessing

## Objective
To clean and prepare the mobile app user survey dataset for Exploratory Data Analysis (EDA), Natural Language Processing (NLP), and Machine Learning.

## Tasks
1. Load the original dataset.
2. Inspect missing values and duplicate records.
3. Analyze response status.
4. Clean column names and text fields.
5. Identify data types and inconsistencies.
6. Prepare a cleaned dataset.
7. Export the processed dataset and cleaning reports.

```python
import pandas as pd
import numpy as np
import os
import re

from google.colab import drive
drive.mount('/content/drive')

PROJECT_PATH = "/content/drive/MyDrive/App-Usage-Intelligence"

DATA_PATH = os.path.join(PROJECT_PATH, "data", "raw")
PROCESSED_PATH = os.path.join(PROJECT_PATH, "data", "processed")
REPORT_PATH = os.path.join(PROJECT_PATH, "reports")

os.makedirs(PROCESSED_PATH, exist_ok=True)
os.makedirs(REPORT_PATH, exist_ok=True)

print("Libraries imported successfully!")
print("Google Drive connected!")
```

```python
file_path = os.path.join(DATA_PATH, "mobile_app_user_dataset.xlsx")

raw = pd.read_excel(file_path, header=None)

columns = raw.iloc[0].astype(str).str.strip().tolist()
descriptions = raw.iloc[1].fillna("").astype(str).tolist()

df = raw.iloc[2:].copy()
df.columns = columns
df.reset_index(drop=True, inplace=True)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python
cleaned_df = df.copy()

print("Original dataset shape:", df.shape)
print("Working dataset shape:", cleaned_df.shape)
```

```python
missing_values = cleaned_df.isnull().sum()

missing_report = pd.DataFrame({
    'Column': missing_values.index,
    'Missing_Count': missing_values.values,
    'Missing_Percentage': (missing_values.values / len(cleaned_df)) * 100
})

missing_report = missing_report.sort_values(
    by='Missing_Percentage',
    ascending=False
)

missing_report.head(20)
```

```python
missing_report.to_csv(
    os.path.join(REPORT_PATH, "missing_values_report.csv"),
    index=False
)
```

```python
duplicate_count = cleaned_df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
duplicate_rows = cleaned_df[cleaned_df.duplicated(keep=False)]

print("Duplicate records found:", len(duplicate_rows))

duplicate_rows.head(10)
```

```python
duplicate_rows.to_csv(
    os.path.join(REPORT_PATH, "duplicate_records_report.csv"),
    index=False
)

print("Duplicate report saved successfully!")
```
